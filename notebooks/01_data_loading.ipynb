Issuer name missingness is systematic rather than random, as all null issuer values are consistently associated with the same BINs. This suggests a BIN enrichment or reference-mapping limitation, rather than incomplete transaction-level data.

#### bin

```python
df.query('bin.isna()')
```

💡

A subset of declined ecommerce transactions consistently presents null BIN and issuer information, while still containing Visa or Mastercard scheme identification, supplied CVC data, no AVS/address data, and the acquirer response ‘invalid card number’.

This pattern suggests that these payments fail during the initial card validation stage, before reaching issuer-level authorization or any downstream optimization opportunities. The absence of BIN and issuer data indicates that the submitted card number could not be successfully validated or identified.

The most likely drivers are shopper card-entry errors, such as mistyped or incomplete card numbers, and potential automated card testing activity, where invalid card credentials are systematically submitted in ecommerce environments. Because these transactions fail prior to authorization processing, they are not recoverable through authorization optimization strategies such as routing or retry logic.


### count

```python
def summarize_columns(df, top_n=10):
    # filter
    cols = df.select_dtypes(include=['object', 'category']).columns

    for col in cols:
        print(f"📌 {col.upper()}")
        counts = df[col].value_counts().head(top_n)
        for val, count in counts.items():
            print(f"  - {val}: {count}")
        print("-" * 30)

summarize_columns(df)
```
